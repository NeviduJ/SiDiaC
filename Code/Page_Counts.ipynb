```python
import os
from pypdf import PdfReader
from collections import Counter
import pandas as pd
```

```python
def get_pdf_details_dataframe(folder_path="data"):
    """
    Loads PDF files from a specified folder, counts the number of pages
    and extracts the title (from filename) for each document,
    returning a pandas DataFrame.

    Args:
        folder_path (str): The path to the folder containing PDF files.
                           Defaults to 'data'.

    Returns:
        pd.DataFrame: A DataFrame with columns 'Title' and 'Pages' for
                      each successfully processed PDF. Returns an empty
                      DataFrame if no PDFs are found or all fail.
    """
    pdf_data = [] # List to store dictionaries of {'Title': ..., 'Pages': ...}

    if not os.path.isdir(folder_path):
        print(f"Error: Folder '{folder_path}' not found.")
        print("Please ensure the 'data' folder exists and contains your PDF files.")
        return pd.DataFrame(columns=['title', 'pages']) # Return empty DataFrame on folder error

    print(f"Processing PDF files in: {folder_path}\n")
    found_pdfs = False
    for filename in os.listdir(folder_path):
        if filename.lower().endswith(".pdf"):
            found_pdfs = True
            filepath = os.path.join(folder_path, filename)
            
            # Extract title from filename (e.g., "book.pdf" -> "book")
            title = os.path.splitext(filename)[0] 

            try:
                reader = PdfReader(filepath)
                num_pages = len(reader.pages)
                pdf_data.append({'title': title, 'pages': num_pages})
                print(f"Processed: {filename} - Title: '{title}', Pages: {num_pages}")
            except Exception as e:
                print(f"Error processing {filename} (Title: '{title}'): {e}")
                # Optionally, you could append an entry with error info:
                # pdf_data.append({'Title': title, 'Pages': 'Error', 'Error_Message': str(e)})
    
    if not found_pdfs:
        print("No PDF files found in the specified folder.")
        return pd.DataFrame(columns=['title', 'pages']) # Return empty DataFrame if no PDFs

    # Create DataFrame from collected data
    df_pdf_details = pd.DataFrame(pdf_data)
    
    if not df_pdf_details.empty:
        # Sort by title for consistent output
        df_pdf_details = df_pdf_details.sort_values(by='title').reset_index(drop=True)
    
    return df_pdf_details
```

```python
if __name__ == "__main__":

    pdf_details_df = get_pdf_details_dataframe(folder_path="Final_Corpus")

    if pdf_details_df is not None and not pdf_details_df.empty:
        print("\n--- PDF Details DataFrame ---")
        print(pdf_details_df.to_markdown(index=False)) # Use to_markdown for nice display
    else:
        print("\nNo PDF details could be retrieved or processed.")
```
